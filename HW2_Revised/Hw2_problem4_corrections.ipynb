```python
import numpy as np
import matplotlib.pyplot as plt
#schehter function is defined as phi(L) = (phi_star/L_star)*(L/L_star)**a *exp(-L/L_star)

# Parameters
a_vals = [0, -1, -1.5]
L_star = 10**10
phi_star = 1e-3

def log10_phi(L, a):
    # The Schechter Function calculation
    phi = (phi_star / L_star) * (L / L_star)**a * np.exp(-L / L_star)
    return np.log10(phi)
    
# Luminosity range (logarithmic space)
x = np.linspace(6, 12, 100)   # This is log10(L)
L = 10**x                    # This is the actual Luminosity

plt.figure(figsize=(10,6))

for a in a_vals:
    plt.plot(
        x,                    # FIX: Plot against 'x' (the log values), not 'L'
        log10_phi(L, a),
        label=f'alpha = {a}'
    )

# Formatting to match the expected scientific output
plt.xlabel(r'log$_{10}(L / L_\odot)$')
plt.ylabel(r'log$_{10}(\Phi)$')
plt.ylim(-30,0)
plt.title('Schechter Luminosity Function')
plt.legend()
plt.grid(True)
plt.savefig("Astr797_Hw2_problem4a")
plt.show()
```

```python
#Problem 4b
import numpy as np
import matplotlib.pyplot as plt
from scipy.special import gamma, gammaincc

#total luminosity is found as p(>L) = phi_star * L_star *gamma(a+2, L/L_star)

for a in a_vals:
    s = a + 2
    y = L/L_star

    # incomplete gamma function
    g =  gammaincc(s, y) * gamma(s)

    rho_L = phi_star * L_star * g
    plt.plot(x, rho_L, label=f'alpha = {a}')

plt.xlabel('log10(L/L_star)')
plt.ylabel(r'$\rho_L(>L)$')
plt.title('Integrated Luminosity Density Above L')
plt.legend()
plt.savefig("Astr797_Hw2_4b")
```
